````python
import numpy as np
````

---
## 📖 Concept 1: Element-wise vs Matrix Multiplication

**CRITICAL DISTINCTION:**
- `*` = element-wise multiplication
- `np.dot()` or `@` = matrix multiplication (dot product)

````python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("A:\n", A)
print("\nB:\n", B)

# Element-wise multiplication
print("\nA * B (element-wise):\n", A * B)

# Matrix multiplication
print("\nA @ B (matrix multiplication):\n", A @ B)
print("\nA.dot(B) (same as A @ B):\n", np.dot(A, B))
````

---
## 📖 Concept 2: Matrix Multiplication Shape Rules

**THE GOLDEN RULE:**

```
(m, n) @ (n, p) → (m, p)
```

The **inner dimensions must match**. The result shape is the **outer dimensions**.

````python
# Valid multiplication
A = np.random.rand(2, 3)  # (2, 3)
B = np.random.rand(3, 4)  # (3, 4)
C = A @ B                 # (2, 4)

print("A shape:", A.shape)
print("B shape:", B.shape)
print("C = A @ B shape:", C.shape)
print("\nRule: (2, 3) @ (3, 4) → (2, 4) ✓")
````

---
## 📖 Concept 3: When Shapes Are Incompatible

If inner dimensions don't match, you get an error.

````python
# This will fail
A = np.random.rand(2, 3)  # (2, 3)
B = np.random.rand(2, 4)  # (2, 4)

try:
    C = A @ B
except ValueError as e:
    print("Error:", e)
    print("\nWhy? (2, 3) @ (2, 4) → Inner dimensions don't match (3 ≠ 2)")
````

---
## 📖 Concept 4: Matrix Multiplication Examples

Let's see the three ways to multiply matrices.

````python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

# Method 1: @ operator (most common)
result1 = A @ B

# Method 2: np.dot()
result2 = np.dot(A, B)

# Method 3: .dot() method
result3 = A.dot(B)

print("Method 1 (A @ B):\n", result1)
print("\nMethod 2 (np.dot(A, B)):\n", result2)
print("\nMethod 3 (A.dot(B)):\n", result3)
print("\nAll three are identical!")
````

### Task 4: Create a chain of matrix multiplications (like a neural network)

````python
# Simulate a simple neural network forward pass
# Input: 1 sample with 4 features
X = np.random.rand(1, 4)

# Layer 1 weights: 4 inputs → 8 neurons
W1 = np.random.rand(4, 8)

# Layer 2 weights: 8 neurons → 3 outputs
W2 = np.random.rand(8, 3)

# Your code here
# Compute: X @ W1 @ W2
````

### ✅ Solution

````python
# Solution
X = np.random.rand(1, 4)
W1 = np.random.rand(4, 8)
W2 = np.random.rand(8, 3)

# Forward pass
hidden = X @ W1    # (1, 4) @ (4, 8) → (1, 8)
output = hidden @ W2  # (1, 8) @ (8, 3) → (1, 3)

print("Input shape:", X.shape)
print("Hidden shape:", hidden.shape)
print("Output shape:", output.shape)
print("\nOutput values:\n", output)
````
