---
## 📖 Concept 1: Creating 1D Arrays

The simplest NumPy array is a 1D array — think of it as a list of numbers.

```python
# From a Python list
arr_1d = np.array([1, 2, 3, 4, 5])
print("Array:", arr_1d)
print("Shape:", arr_1d.shape)  # (5,) means 1D array with 5 elements
print("Dimensions:", arr_1d.ndim)  # 1 dimension
```

---
## 📖 Concept 2: Creating 2D Arrays (Matrices)

A 2D array is created from nested lists. Think of it as rows and columns.

```python
# 2D array (matrix)
arr_2d = np.array([[1, 2, 3],
                   [4, 5, 6]])
print("Array:\n", arr_2d)
print("Shape:", arr_2d.shape)  # (2, 3) means 2 rows, 3 columns
print("Dimensions:", arr_2d.ndim)  # 2 dimensions
```

---
## 📖 Concept 3: Inspecting Arrays

Key attributes to understand any array:

```python
arr = np.array([[1.5, 2.3, 3.7],
                [4.1, 5.9, 6.2]])

print("Shape:", arr.shape)      # (rows, columns)
print("Dimensions:", arr.ndim)  # number of axes
print("Data type:", arr.dtype)  # float64, int32, etc.
print("Total size:", arr.size)  # total number of elements (2*3 = 6)
```

---
## 📖 Concept 4: Array Creation Functions

NumPy has convenient functions for creating arrays.

```python
# Create array of zeros
zeros = np.zeros((3, 4))  # 3 rows, 4 columns of zeros
print("Zeros:\n", zeros)

# Create array of ones
ones = np.ones((2, 3))  # 2 rows, 3 columns of ones
print("\nOnes:\n", ones)

# Create range of numbers
range_arr = np.arange(0, 10, 2)  # start, stop, step
print("\nRange:", range_arr)

# Create evenly spaced numbers
linspace_arr = np.linspace(0, 1, 5)  # 5 numbers from 0 to 1
print("\nLinspace:", linspace_arr)

# Create random numbers
random_arr = np.random.rand(2, 3)  # 2x3 matrix of random numbers [0, 1)
print("\nRandom:\n", random_arr)
```
