# 06 - Views vs Copies

## 🎯 Objective
**Understand NumPy side effects** — this is CRITICAL to avoid silent bugs.

## 🤖 AI Relevance
**Silent bugs in ML pipelines come from this.** Imagine:
- Splitting data into train/test, but accidentally modifying both
- Normalizing a subset, but changing the original dataset
- Creating "different" batches that share memory

**Understanding views vs copies prevents production disasters.**

```python
import numpy as np
```

---
## 📖 Concept 1: Assignment Creates Views (Not Copies!)

**SURPRISE:** When you do `B = A`, B is just another name for A. They share memory!

```python
A = np.array([1, 2, 3, 4, 5])
B = A  # This does NOT copy!

print("A:", A)
print("B:", B)

# Modify B
B[0] = 999

print("\nAfter modifying B[0] = 999:")
print("A:", A)  # A also changed!
print("B:", B)
```

---
## 📖 Concept 2: Slicing Creates Views

**Another surprise:** Slices are views, not copies!

```python
A = np.array([1, 2, 3, 4, 5])
B = A[1:4]  # Slice is a view!

print("A:", A)
print("B:", B)

# Modify B
B[0] = 999

print("\nAfter modifying B[0] = 999:")
print("A:", A)  # A changed! The slice was a view.
print("B:", B)
```

---
## 📖 Concept 3: Using .copy() to Create Independent Copies

**The solution:** Use `.copy()` when you want a true independent copy.

```python
A = np.array([1, 2, 3, 4, 5])
B = A[1:4].copy()  # Now B is an independent copy

print("A:", A)
print("B:", B)

# Modify B
B[0] = 999

print("\nAfter modifying B[0] = 999:")
print("A:", A)  # A unchanged! B was a copy.
print("B:", B)
```

---
## 📖 Concept 4: Checking if Arrays Share Memory

Use `np.shares_memory()` to check if two arrays share the same memory.

```python
A = np.array([1, 2, 3, 4, 5])

# View (shares memory)
B = A[1:4]
print("B = A[1:4]")
print("Do A and B share memory?", np.shares_memory(A, B))  # True

# Copy (independent)
C = A[1:4].copy()
print("\nC = A[1:4].copy()")
print("Do A and C share memory?", np.shares_memory(A, C))  # False
```

---
## 📖 Concept 5: When Operations Create Copies

**Operations that create new arrays (copies):**
- Arithmetic operations: `A + 1`, `A * 2`
- Boolean indexing: `A[A > 3]`
- Concatenation: `np.concatenate()`, `np.vstack()`, etc.

**Operations that create views:**
- Slicing: `A[1:3]`
- Reshaping (usually): `A.reshape()`
- Transpose: `A.T`

```python
A = np.array([1, 2, 3, 4, 5])

# Arithmetic creates a copy
B = A + 1
print("B = A + 1")
print("Shares memory?", np.shares_memory(A, B))  # False

# Boolean indexing creates a copy
C = A[A > 3]
print("\nC = A[A > 3]")
print("Shares memory?", np.shares_memory(A, C))  # False

# Slicing creates a view
D = A[1:3]
print("\nD = A[1:3]")
print("Shares memory?", np.shares_memory(A, D))  # True
```

## 🎯 Key Takeaways

1. **Assignment = view** — `B = A` doesn't copy, they share memory
2. **Slicing = view** — `B = A[1:3]` is a view, not a copy
3. **Use .copy()** — When you want independence: `B = A.copy()`
4. **Check with np.shares_memory()** — Use this to debug
5. **Arithmetic creates copies** — `A + 1`, `A * 2` create new arrays
6. **Boolean indexing creates copies** — `A[A > 5]` creates new array

**Remember:** When in doubt, use `.copy()`. It's better to be safe than to debug silent bugs in production! 🚀
